SUBSCRIBERS TABLE CLEANING

```python
subscribers['is_active'] = subscribers['is_active'].astype(str).str.strip().str.upper().map({
    'TRUE':'Active',
    'FALSE':'Inactive',
    '1':'Active',
    '0':'Inactive'
})
```
